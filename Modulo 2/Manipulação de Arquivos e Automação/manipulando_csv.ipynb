# CSV


CSV (Comma-Separated Values) é um formato de arquivo de texto simples usado para armazenar dados em formato de tabela, onde:
- cada linha represeta um registro
- cada coluna é separada por vírgulas (ou outro delimitador)
- Pode ser aberto por Excel , GoogleSheets,pandas etc

Exemplo de conteúdo de um alunos.csv:
nome,idade,curso
Maria, 30, Python
João, 25, Data Science

O módulo csv permite ler e escrever arquivos CSV facilmente. Ele oferece duas formas principais:
- usando listas como csv.reader() e csv.writer()
- usando dicionarios com csv.DicReader() e csv.DirWriter()

```python
import csv
```

```python
# 1 Criar um csv com csv.write()
dados = [
    ['nome', 'idade', 'curso'],
    ['Maria', 30, 'Python'],
    ['João', 25, 'Data Science']
]

# Abrir o arquivo em modo escrita ('w'), com encoding e newline=''ArithmeticError
with open('alunos.csv','w',newline='',encoding='utf=8') as arquivos:
    # criamos um objeto 'write' que é o escritor de um arquivo csv.
    # Ele é responsável por transformar listas em linhas no arquivo csv
    escritor = csv.writer(arquivos)
    # Escrever todas as linhas as lista 'dados' de uma vez, .writerows é um método do objeto escritor
    escritor.writerows(dados)
    
```

Observação sobre o uso de newline="Sempre use newline="ao abrir o arquivo com open() para evitar quebrar de linhas duplicadas, principalmente no Windows

# 2. Ler um CSV com csv.reader()

```python
# Abrimos o arquivo em modo leitura('r'), com endoding
with open('alunos.csv', 'r', encoding='utf-8') as arquivo:
    # o reader cria um objeto iteravél que transforma cada linha do CSV em uma lista.
    leitor= csv.reader(arquivo)

    for linha in leitor:
        print(linha)
```

# 3. Lendo como dicionario com csv.DictReader()


```python
# Abrimos o arquivo em modo leitura('r'), com endoding
with open('alunos.csv', 'r', encoding='utf-8') as arquivo:
    # o reader cria um objeto iteravél que transforma cada linha do CSV em uma lista.
    leitor= csv.DictReader(arquivo)

    for linha in leitor:
        print(linha)
```

# 4. Adicionando uma linha ao csv


```python
import csv  # Certifique-se de importar o módulo csv

nova_linha = ['Fernando', 27, 'Banco de Dados']

with open('alunos.csv', 'a', newline='', encoding='utf-8') as arquivos:
    escritor = csv.writer(arquivos)  # Usando a variável correta 'arquivos'
    escritor.writerow(nova_linha)  # Usando writerow para escrever uma única linha
```
