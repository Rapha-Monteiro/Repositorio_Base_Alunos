# Lendo um txt


Em Python, usamos a função open()para abrir arquivos


```python
arquivo = open('hellow_world.txt','r')
print(arquivo)
```

Os modos são:
- r - leitura(read)
- w - escrita(write, sobrescreve)
- a - escrita(append, adiciona ao final)
- x - escrita(cria um novo arquivo, erro se ja existir)
- b - modo binario
- t - modo texto(padrão)


# Lendo um arquivo linha por linha


```python
with open('hello_world.txt','r') as f:
    for linha in f:
        print(linha)
```

O objeto _io.textIOWrapper é a classe que representa arquivos abertos em modo texto em Python. Sempre que você abre um arquivo usando open() com o modo como r, w, a (ou seja, sem o modo binario, b), o que você recebe é uma instância de _io.TextIOWrapper
