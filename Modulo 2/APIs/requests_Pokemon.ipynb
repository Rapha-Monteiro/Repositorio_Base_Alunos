# API Pokemon


```python
import requests
res = requests.get('https://pokeapi.co/api/v2/pokemon/pikachu')
print(res.text)
print(type(res.text))
dados_pokemon = res.json()
print(type(dados_pokemon))
```

```python
# Verificamos as chaves do dicionario
for chave in dados_pokemon.keys():
    print(chave)
```

```python
# Acessando dados do dicionario 
print(dados_pokemon['name'])
print(dados_pokemon['moves'])
print(dados_pokemon['types'])
```

```python
# Customizr a busca pelo nome
import requests
pokemon= input('Digite o nome de um Pokemon: ').lower()
url='https://pokeapi.co/api/v2/pokemon/{pokemon}'
res = requests.get(url)

if res.status_code == 200:
	data = res.json()
	print(f'Nome:{data['name']}')
	print('Tipos: ',','.join(t['type']['name']for t in data['types']))
else:
	print('Pokémon não encontrado')
```
